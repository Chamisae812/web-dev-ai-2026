### 데이터베이스와 테이블 만들기

```python
import pymysql

conn = pymysql.connect(host="127.0.0.1", user="root", password="qwer1234")
cur = conn.cursor()
cur.execute("CREATE DATABASE IF NOT EXISTS test_db")
conn.close()
```

```python
conn = pymysql.connect(host="127.0.0.1", user="root", password="qwer1234", db="test_db")
cur = conn.cursor()
cur.execute("""
CREATE TABLE IF NOT EXISTS member(
    id INT AUTO_INCREMENT PRIMARY KEY,
    name VARCHAR(255),
    address VARCHAR(255),
    email VARCHAR(255),
    phone_number VARCHAR(255),
    job VARCHAR(255),
    company VARCHAR(255)
    )
    """)
conn.commit()
print("테이블 생성 완료")
```

```python
from faker import Faker
fake = Faker('ko-KR')
print(fake.name())
print(fake.address())
print(fake.email())
print(fake.phone_number())
print(fake.company())
print(fake.job())
```

### CRUD - INSERT
- 데이터 추가

```python
cur.execute("""
    INSERT INTO member (name, address, email, phone_number, company, job)
    VALUES (%s, %s, %s, %s, %s, %s)
    """, (fake.name(), fake.address(), fake.email(), fake.phone_number(), fake.company(), fake.job()))
conn.commit
print("저장")
```

```python
for _ in range(10):
    cur.execute("""
    INSERT INTO member (name, address, email, phone_number, company, job)
    VALUES (%s, %s, %s, %s, %s, %s)
    """, (fake.name(), fake.address(), fake.email(), fake.phone_number(), fake.company(), fake.job()))
    conn.commit
```

```python
# excutemany: INSERT문 하나에 데이터 목록을 통째로 넘기면, 반복문 없이 여러 행을 한번에 저장 가능
members = [(fake.name(), fake.address(), fake.email(), fake.phone_number(), fake.company(), fake.job())
           for _ in range(10)]
members

cur.executemany("""
INSERT INTO member (name, address, email, phone_number, company, job)
VALUES (%s, %s, %s, %s, %s, %s)
""", members)
conn.commit

print("저장")
```

### CRUD - SELECT
- 데이터 조회

```python
cur.execute("SELECT * FROM member")
for row in cur.fetchall(): #전체 조회
    print(row)
```

```python
cur.execute("SELECT * FROM member WHERE id = 1")
cur.fetchone() # 결과 조건이 1개
```

### CRUD _ UPDATE
- 데이터 수정

```python
cur.execute("UPDATE member SET address = %s WHERE id = %s", ('서울특별시 강남구 테헤란로 123', 1))
conn.commit()
```

### CRUD - DELETE
- 데이터 삭제

```python
cur.execute("DELETE FROM member WHERE id = %s", 1)
conn.commit()
```
